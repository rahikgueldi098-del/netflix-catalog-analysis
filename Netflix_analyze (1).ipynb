```python
import pandas as pd

df = pd.read_csv("netflix data.csv")
print(df.shape)
df.head()
```

```python
print(list(df.columns))
print()
print(df.isna().sum().sum(), "missing values in total")
```

```python
print(df.isna().sum())
print()
df.info()
```

```python
df = df.dropna(subset=["type"])

for col in ["title", "listed_in", "rating", "unit"]:
    df[col] = df[col].fillna("Unknown")

print(df.isna().sum())
```

```python
print(df["type"].value_counts())
```
